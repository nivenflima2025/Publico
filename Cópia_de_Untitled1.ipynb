```python
import tkinter as tk
from tkinter import messagebox

class Livro:
    def __init__(self, titulo, autor, livro_id):
        self.titulo = titulo
        self.autor = autor
        self.id = livro_id
        self.emprestado = False

    def __str__(self):
        status = "Emprestado" if self.emprestado else "Disponível"
        return f"{self.id} - {self.titulo} ({self.autor}) - {status}"

class Membro:
    def __init__(self, nome, membro_id):
        self.nome = nome
        self.id = membro_id
        self.historico = []

    def emprestar_livro(self, livro):
        self.historico.append(livro)

    def __str__(self):
        return f"{self.id} - {self.nome}"

class Biblioteca:
    def __init__(self):
        self.livros = []
        self.membros = []

    def adicionar_livro(self, livro):
        self.livros.append(livro)

    def adicionar_membro(self, membro):
        self.membros.append(membro)

    def buscar_livro(self, termo):
        return [livro for livro in self.livros if termo.lower() in livro.titulo.lower() or termo.lower() in livro.autor.lower() or termo == livro.id]
def emprestar_livro(self, livro_id, membro_id):
        livro = next((l for l in self.livros if l.id == livro_id), None)
        membro = next((m for m in self.membros if m.id == membro_id), None)
        if livro and not livro.emprestado:
            livro.emprestado = True
            membro.emprestar_livro(livro)
            return True
        return False

    def devolver_livro(self, livro_id):
        livro = next((l for l in self.livros if l.id == livro_id), None)
        if livro and livro.emprestado:
            livro.emprestado = False
            return True
        return False

class InterfaceBiblioteca:
    def __init__(self, root):
        self.bib = Biblioteca()
        self.root = root
        root.title("Biblioteca")

        self.frame = tk.Frame(root)
        self.frame.pack(padx=10, pady=10)

        tk.Label(self.frame, text="Título:").grid(row=0, column=0)
        self.titulo_entry = tk.Entry(self.frame)
        self.titulo_entry.grid(row=0, column=1)

        tk.Label(self.frame, text="Autor:").grid(row=1, column=0)
        self.autor_entry = tk.Entry(self.frame)
        self.autor_entry.grid(row=1, column=1)

        tk.Label(self.frame, text="ID:").grid(row=2, column=0)
        self.id_entry = tk.Entry(self.frame)
        self.id_entry.grid(row=2, column=1)

        tk.Button(self.frame, text="Adicionar Livro", command=self.adicionar_livro).grid(row=3, columnspan=2, pady=5)

        tk.Label(self.frame, text="Nome Membro:").grid(row=4, column=0)
        self.nome_membro_entry = tk.Entry(self.frame)
        self.nome_membro_entry.grid(row=4, column=1)

        tk.Label(self.frame, text="ID Membro:").grid(row=5, column=0)
        self.id_membro_entry = tk.Entry(self.frame)
        self.id_membro_entry.grid(row=5, column=1)

        tk.Button(self.frame, text="Adicionar Membro", command=self.adicionar_membro).grid(row=6, columnspan=2, pady=5)


        tk.Button(self.frame, text="Emprestar Livro", command=self.emprestar_livro).grid(row=7, columnspan=2, pady=5)
        tk.Button(self.frame, text="Devolver Livro", command=self.devolver_livro).grid(row=8, columnspan=2, pady=5)

        tk.Label(self.frame, text="Buscar Livro:").grid(row=9, column=0)
        self.busca_entry = tk.Entry(self.frame)
        self.busca_entry.grid(row=9, column=1)
        tk.Button(self.frame, text="Buscar", command=self.buscar_livro).grid(row=10, columnspan=2)

        self.resultado = tk.Text(self.frame, height=10, width=50)
        self.resultado.grid(row=11, columnspan=2, pady=10)

    def adicionar_livro(self):
        titulo = self.titulo_entry.get()
        autor = self.autor_entry.get()
        livro_id = self.id_entry.get()
        if titulo and autor and livro_id:
            self.bib.adicionar_livro(Livro(titulo, autor, livro_id))
            messagebox.showinfo("Sucesso", "Livro adicionado com sucesso!")
        else:
            messagebox.showwarning("Erro", "Preencha todos os campos do livro.")

    def adicionar_membro(self):
        nome = self.nome_membro_entry.get()
        membro_id = self.id_membro_entry.get()
        if nome and membro_id:
            self.bib.adicionar_membro(Membro(nome, membro_id))
            messagebox.showinfo("Sucesso", "Membro adicionado com sucesso!")
        else:
            messagebox.showwarning("Erro", "Preencha todos os campos do membro.")

    def emprestar_livro(self):
        livro_id = self.id_entry.get()
        membro_id = self.id_membro_entry.get()
        if self.bib.emprestar_livro(livro_id, membro_id):
            messagebox.showinfo("Sucesso", "Livro emprestado com sucesso!")
        else:
            messagebox.showwarning("Erro", "Livro não encontrado ou já emprestado.")

    def devolver_livro(self):
        livro_id = self.id_entry.get()
        if self.bib.devolver_livro(livro_id):
            messagebox.showinfo("Sucesso", "Livro devolvido com sucesso!")
        else:
            messagebox.showwarning("Erro", "Livro não está emprestado ou não encontrado.")

    def buscar_livro(self):
        termo = self.busca_entry.get()
        resultados = self.bib.buscar_livro(termo)
        self.resultado.delete(1.0, tk.END)
        if resultados:
            for livro in resultados:
                self.resultado.insert(tk.END, str(livro) + "\n")
        else:
            self.resultado.insert(tk.END, "Nenhum livro encontrado.")






```
